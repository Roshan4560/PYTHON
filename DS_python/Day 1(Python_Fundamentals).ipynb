```python
# implicit conversion

a = 0b1010  # Binary Literals
b = 100     # Decimal Literals
c = 0o310   # Octal Literal
d = 0x12c   # hexadecimal literals

print(a)
print(b)
print(c)
print(d)
```

```python
# and operator -> isme agar dono true hai tabhi true dega warna false return karega.
# ye sirf 0 ko false manta hai baki sabhi ko True manta hai.

print(1 and 0)
print(1 and 2)
```

```python
# or operator -> isme agar dono false hai tabhi false dega warna true return karega.

print(1 or 0)
```

```python
# not -> isme agar true hai to false return aur false hai to true return karega

print(not 1)
print(not 0)
```
