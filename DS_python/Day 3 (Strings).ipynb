```python
# check palidrome or not
# eg = malayalam, abba

s = input("Enter the strings:")
flag = True

for i in range(0,len(s)//2):
    if s[i]!=s[len(s)-i-1]:
        flag = False
        break
if flag:
    print("Palidrome")
else:
    print("Not palidrome")
```
