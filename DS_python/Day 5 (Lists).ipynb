```python
l = [1,2,3,4,55,6,77]
print(l[5])
print(l[-3])
print(l[0::2])
print(l[::-1])
```

```python
l = [[1,2,3],[4,5,6]]
print(l[0][2])
```

```python
# append() = single item added

l = [1,2,3,4,5]
l.append(True)
print(l)
```

```python
# extend() = multiple item added

l = [1,2,3,4,5]
l.extend([6,7,8])
print(l)
```

```python
# insert = at specific position item is added

l = [1,2,3,4,5,6,7]
l.insert(2,1234)
print(l)
```

```python
l = [1,2,3,4,5]
l[-1] = 500
print(l)
```

```python
# del

l = [1,2,3,4]
print(l)      # print the list
del(l)
print(l)     # doesnot print it because delete it
```

```python
# remove

l = [1,23,3,4,54,5,6,7]

l.remove(54)
print(l)
```

```python
# pop()

l = [1,2,3,4,5,6,7,8]
l.pop(4)                  # pop out the 5
print(l)
l.pop()
print(l)               # pop out the last item 
```

```python
# clear()

l = [1,2,3,4]
l.clear()
print(l)      # blank list is created
```

```python
# arithmatic  (+,*)

l1 = [1,2,3,4]
l2 = [5,6,7,8,9]

print(l1+l2)     # concatenation
```

```python
print(l1*3)
```

```python
# membership

l1 = [1,2,3,4,5]
l2 = [1,2,3,4,[5,6]]

print(5 in l1)        # 5 present in l1 -> true
print(4 not in l2)    # 4 not in l2 -> False
print(5 in l2)        # 5 in l2 -> False
```

```python
for i in [1,2,3,4,5]:
    print(i)
```

```python
# len()

l = [1,2,3,4,5,6,7]
len(l)
```

```python
# min()

l = [1,2,3,4,5,6,7]
min(l)
```

```python
# max()

l = [1,2,3,4,5,6,7]
max(l)
```

```python
# sorted()

l = [1,4,6,8,3,2,5,7]

sorted(l)
```

```python
sorted(l,reverse=True)
```

```python
# count()

l = [1,2,3,4,5,6,7]
l.count(5)
```

```python
# reverse()

l = [1,2,3,4,5,6,7]
l.reverse()         # permananet reverse the lists
print(l)
```

```python
# index()

l = [1,2,3,4,5,6,7]
l.index(4)
```

```python
# copy()

l = [2,5,1,3,4,8,6,7]
l1 = l.copy()
print(l1)
```

```python
# sort()

l = [2,5,1,3,4,8,6,7]

print(l)
print(sorted(l))     # temporary sorted
print(l)
l.sort()            # permananent sort
print(l)
```
