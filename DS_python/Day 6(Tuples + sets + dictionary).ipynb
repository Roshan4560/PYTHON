# Sets

```python
s = set()
print(type(s))

s = {1,2,3,3,4,5,6,6,8}
print(s)
print(type(s))
```

```python
s = {1,2,3,'kvslvnvls',(2,6,7,8),True}

print(s)
```

```python
# add()

s= {1,2,3,4,5}
s.add(6)
print(s)
```

```python
# update()

s.update([7,8,9,10])
print(s)
```

```python
# del()

s = {1,2,3,4,5}
print(s)
del s
print(s)
```

```python
# discard()

s = {1,2,3,4,5,6}
s.discard(6)
print(s)
```

```python
# remove()

s = {1,2,3,4,5}
s.remove(5)
print(s)
```

```python
# union(|)

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}

s1|s2
```

```python
# intersection(&)

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}
s1&s2
```

```python
# difference(-)

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}
print(s1-s2)
print(s2-s1)
```

```python
# symmetric difference(^)

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}
s1^s2
```

```python
# membership test

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}
print(1 in s1)
print(2 not in s1)
```

```python
# iteration

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}
for i in s1:
    print(i)
```

```python
# len()

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}
print(len(s1))
print(len(s2))
```

```python
# sum()

print(sum(s1))
print(sum(s2))
```

```python
# min()

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}
print(min(s1))
print(min(s2))
```

```python
# max()

s1 = {1,2,3,4,5,6}
s2 = {5,6,7,8,9,10}
print(max(s1))
print(max(s2))
```

```python
# sorted()

s1 = {2,1,5,4,3,6}
s2 = {10,8,7,6,9,5}
print(sorted(s1))
```

```python
# isdisjoint  means koi common nhi hona chahiyee

s1 = {1,2,3,4}
s2 = {7,8,9,10}
s1.isdisjoint(s2)
```

```python
# issubset()  to check that subset hai ya nhi

s1 = {1,2,3,4,5,6,7,8}
s2 = {1,2,3,4}
s2.issubset(s1)
```

```python
# set comprehension

{i**2 for i in range(1,11)}
```
