# Functions

```python
# number is even or odd

def iseven(n):
    if n%2==0:
        return 'even'
    else:
        return 'odd'

print(iseven(20))
```

```python
# Default arguments 

def power(a=1,b=1):
    return a**b


print(power())  # default arguments hai,  matlab kuch na do to default me a=1,b=1 bheja
```

```python
power(2)
```

```python
# keyword arguments

power(b=3,a=2)   # hm a ko baad me aur b ko phele bhej sakte hai
```

# *args and **kwargs

```python
# *args -> allows us to pass a variable number of non-keyboard arguments to a functions 

def multiply(*args):   # *args help to input variable inputs 
    product = 1

    for i in args:
        product = product*i

    return product
```

```python
multiply(1,2,3,4,5,6)
```

```python
# **kwargs -> **kwargs allows us to pass any number of keyword arguments.
# keyword arguments mean that they contain a key-value pair,like a python dictionary.

def display(**kwargs):
    for (key,value) in kwargs.items():
        print(key,'->',value)
```

```python
display(India='Delhi',srilanka='colombo',nepal='kathmmandu')
```
